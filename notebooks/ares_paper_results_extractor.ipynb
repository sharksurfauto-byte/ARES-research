```python
# === Display Generated Publication Figures Inline ===
from IPython.display import Image, display
from pathlib import Path

figures_dir = Path("/kaggle/working/ARES-research/paper/figures")
if not figures_dir.exists():
    figures_dir = Path("/kaggle/working/ares_paper_assets/figures")

figs = sorted(list(figures_dir.glob("*.png")))
for f in figs:
    print(f"\n--- {f.name} ---")
    display(Image(filename=str(f), width=750))
```

```python
# === Verify Downloadable Paper Assets Zip ===
import os
zip_path = "/kaggle/working/ares_paper_assets.zip"
if os.path.exists(zip_path):
    size_kb = os.path.getsize(zip_path) / 1024
    print(f"SUCCESS! '{zip_path}' is ready ({size_kb:.1f} KB).")
    print("You can download it directly from the Output pane on the right-hand sidebar!")
else:
    print("Warning: Zip file not found.")
```
